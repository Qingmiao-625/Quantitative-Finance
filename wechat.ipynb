```python
import itchat


itchat.auto_login(hotReload=False)  # 热启动避免重复扫码
itchat.run()
```

```python
@itchat.msg_register(itchat.content.TEXT)
def reply(msg):
    if msg['Text'] == 'hello':
        return 'Hello from Bot!'
    elif msg['Text'] == '你好':
        return '你好，我是机器人！'
```

```python
# 改进好友搜索方式
target_remarkname = "WR"
target_username = "sunnywr98"

# 优先通过备注名搜索
friend = itchat.search_friends(remarkName=target_remarkname)
if friend:
    # 使用UserName作为唯一标识
    itchat.send("你好！我是微信机器人！", toUserName=friend[0]['UserName'])
    print(f"消息已发送给 [{friend[0]['NickName']}]")
else:
    # 备选通过微信号搜索
    friend = itchat.search_friends(userName=target_username)
    if friend:
        itchat.send("你好！我是微信机器人！", toUserName=friend[0]['UserName'])
        print(f"消息已发送给 [{friend[0]['NickName']}]")
    else:
        print(f"未找到备注名为 [{target_remarkname}] 或微信号 [{target_username}] 的好友")

# 添加异常处理
try:
    itchat.run()
except Exception as e:
    print(f"运行出错: {str(e)}")
```
